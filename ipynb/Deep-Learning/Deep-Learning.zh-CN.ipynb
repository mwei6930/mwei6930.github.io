深度学习研究分层参数化系统如何从数据中学习有用表示。完整的知识体系需要把数学运算、优化、架构、表示、生成、可扩展系统和可靠评估连接起来，而不能把各类模型当作彼此孤立的配方。

本系列遵循这些依赖关系展开：第一部分建立训练基础；第二部分研究主要架构家族；第三部分解释可复用表示与基础模型；第四部分连接生成、序列决策和对齐；第五部分讨论可扩展工程、可靠性与研究实践。

## **学习路径**

```text
Tensor 与计算图
-> 神经网络构件与反向传播
-> 优化与泛化
-> 架构家族
-> 表示、迁移与基础模型
-> 生成、决策与对齐
-> 可扩展系统、可靠性与研究
```

## **实现层级**

- **Tier 1：从零实现机制。** 对自动微分、注意力、RoPE、LoRA 和小型生成过程等核心思想进行推导和直接实现。
- **Tier 2：构建组件。** 使用基础 PyTorch 运算组装架构与目标函数，但不重复实现生产级内核。
- **Tier 3：系统实验。** 通过伪代码、显存模型、官方接口与性能分析学习分布式训练、编译和优化内核。
