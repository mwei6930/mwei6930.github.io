Deep learning studies how layered parameterized systems learn useful representations from data. A complete treatment must connect mathematical operations, optimization, architecture, representation, generation, scalable systems, and reliable evaluation rather than treating model families as isolated recipes.

This series follows those dependencies. Part I develops the training foundations, Part II studies the major architecture families, Part III explains reusable representations and foundation models, Part IV connects generation, sequential decisions, and alignment, and Part V addresses scalable engineering, reliability, and research practice.

## **Learning Path**

```text
tensors and computation graphs
-> neural building blocks and backpropagation
-> optimization and generalization
-> architecture families
-> representation, transfer, and foundation models
-> generation, decision making, and alignment
-> scalable systems, reliability, and research
```

## **Implementation Levels**

- **Tier 1: mechanism from scratch.** Core ideas such as automatic differentiation, attention, RoPE, LoRA, and small generative processes are derived and implemented directly.
- **Tier 2: component construction.** Architectures and objectives are assembled from basic PyTorch operations without reproducing production kernels.
- **Tier 3: systems experiment.** Distributed training, compilation, and optimized kernels are studied through pseudocode, memory models, official interfaces, and profiling.
