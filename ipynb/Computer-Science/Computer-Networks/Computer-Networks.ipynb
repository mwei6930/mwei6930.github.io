Computer networks explain how independently administered machines exchange information across unreliable, shared, and heterogeneous infrastructure. A complete understanding requires more than memorizing protocol names: it must connect packet formats and state machines to architectural principles, distributed routing decisions, performance limits, security boundaries, and operational evidence.

This series uses a systems-oriented path. It first follows one packet through the Internet, then studies local delivery, IP forwarding, routing, reliable transport, congestion control, applications, security, operations, and programmable infrastructure. The recurring HTTPS request connects the chapters so that each mechanism remains part of one end-to-end system.

**Learning path**

Internet architecture -> local delivery -> IP forwarding -> routing -> reliable transport -> congestion control -> applications -> security -> operations -> programmable infrastructure

## **Recurring End-to-End Case**

The running case begins when a host joins a network and a user opens an HTTPS URL. The path introduces Wi-Fi or Ethernet access, address configuration, local resolution, DNS, IP forwarding, intra- and interdomain routing, TCP or QUIC, congestion control, TLS, HTTP, content delivery, cloud infrastructure, and operational telemetry. Each chapter expands the relevant segment of this path.

## **Scope Boundaries**

This series focuses on Internet architecture, protocols, implementation mechanisms, performance, security, and operations. Detailed modulation and signal-processing theory belong to digital communications; cryptographic proofs and exploit development belong to security-specialist paths; consensus, replication, and distributed transactions belong to distributed systems. Their network-facing mechanisms remain in scope where they clarify end-to-end behavior.
