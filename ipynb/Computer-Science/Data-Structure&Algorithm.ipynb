Data structures and algorithms focus on a core issue: how to organize information and solve computational problems efficiently. A data structure determines how data is stored, accessed, and updated, while an algorithm describes the steps used to transform an input into a required output. As the input grows, apparently small design decisions can create major differences in running time, memory use, and scalability.

This guideline combines three complementary perspectives. The COMP2123 material establishes algorithm analysis and fundamental data structures; COMP3027 develops algorithm-design techniques, network flow, reductions, and computational hardness; and representative LeetCode Hot 100 problems turn these ideas into reusable problem-solving patterns. The chapters follow their conceptual dependencies rather than listing techniques in isolation.

**Learning path**

analysis and correctness $\rightarrow$ core data structures $\rightarrow$ searching and graph algorithms $\rightarrow$ algorithm-design paradigms $\rightarrow$ advanced algorithms and complexity
