An operating system turns processor, memory, storage, and devices into abstractions that programs can use safely. It must multiplex limited resources, preserve isolation between mutually untrusted programs, coordinate concurrent activity, recover persistent state after failures, and expose interfaces that remain useful across changing hardware.

This series follows the life of a program rather than treating operating-system mechanisms as an unrelated catalog. It begins at the user-kernel boundary, follows process creation and concurrent execution, explains CPU and memory virtualization, traces I/O into persistent storage, and ends with protection, virtual machines, containers, and modern kernel design.

**Learning path**

kernel boundary -> process execution -> threads and IPC -> scheduling -> synchronization -> address translation -> demand paging -> I/O -> file systems -> protection -> virtualization

## **Recurring End-to-End Case**

The running case is the shell pipeline <code>cat input.txt | grep kernel &gt; result.txt</code>. The shell creates processes, connects a pipe, installs file descriptors, loads executable code, and asks the kernel to schedule the resulting threads. Their instructions trigger address translation and page faults; their reads and writes cross the I/O and file-system stacks; permissions constrain access; and a container or virtual machine can add another isolation boundary. Each chapter expands one part of this execution path.

## **Connections to Other Computer Science Paths**

Computer Organization explains the privilege, page-table, interrupt, DMA, cache-coherence, and atomic-instruction mechanisms on which the operating system relies. Computer Networks explains the protocols reached through socket and network-device interfaces. Data Structures and Algorithms supplies the queues, trees, hash tables, graphs, and optimization ideas used inside schedulers, memory managers, and file systems. This series focuses on how the operating system combines those foundations into protected services and policies.

## **Scope Boundaries**

This series focuses on operating-system abstractions, interfaces, implementation mechanisms, policies, correctness, performance, persistence, and isolation. Detailed processor circuit design belongs to Computer Organization; language parsing and optimization belong to compilers; network protocol design belongs to Computer Networks; consensus, replication, and distributed transactions belong to a future Distributed Systems path. Their operating-system-facing mechanisms remain in scope where they clarify a complete execution path.
