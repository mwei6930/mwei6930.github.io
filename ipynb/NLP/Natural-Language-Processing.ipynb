Natural Language Processing (NLP) is a major subfield of AI that focuses on enabling computers to understand, analyze, and generate human language. Unlike structured data, natural language is highly contextual and ambiguous: the same expression may carry different meanings across speakers, domains, and situations. A complete NLP system therefore needs more than a model. It also needs a clear account of language, carefully constructed data, an appropriate input-output formulation, a training strategy, and an evaluation protocol that reflects the intended use.

The chapters below follow the dependency of an NLP project rather than the historical order in which methods appeared. Language and data come first; tokenisation turns text into model units; representation and modeling define what can be learned; training and evaluation determine whether learning succeeds; benchmarks make research claims comparable; tasks connect methods to applications; and the final chapters address risk and complete system design.

```text
language and data
-> tokenisation
-> representation
-> modeling
-> training
-> evaluation and comparison
-> tasks and applications
-> responsible production systems
```
