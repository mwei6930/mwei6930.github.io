### **Probability**
#### **Probability Space**($\Omega$, $\mathcal{F}$, $P$)
In probability theory, a **probability space** provides a formal mathematical framework to describe randomness.
- **Sample Space**($\Omega$): The set of all possible result, all possible basic situations that may occur.
- **Event Domain**($\mathcal{F}$, $\sigma$-algebra): The collection of subsets of $\Omega$ that we can assign probabilities to. It must satisfy the following properties:
    - Contains the sample space: $\Omega \in \mathcal{F}$
    - Closed under complement: $A \in \mathcal{F} \Rightarrow A^c \in \mathcal{F}$
    - Closed under countable unions:$A_1, A_2, \dots \in \mathcal{F} \Rightarrow \bigcup_{i=1}^{\infty} A_i \in \mathcal{F}$
- **Probability Measure**($P$): The **probability measure** assigns a probability to each event: $P: \mathcal{F} \rightarrow [0,1]$. It must satisfy the following axioms (Kolmogorov axioms):
    - Non-negativity: $P(A) \ge 0$
    - Normalization: $P(\Omega) = 1$
    - Countable additivity (for disjoint events):$P\left(\bigcup_{i=1}^{\infty} A_i\right) = \sum_{i=1}^{\infty} P(A_i)$
- For example: Rolling a Die, Sample Space: $\Omega = \{1,2,3,4,5,6\}$, Event Domain: $\mathcal{F} = 2^\Omega$, Let $A = \{2,4,6\}$, Probability Measure: $P(A) = \frac{|A|}{6} = \frac{3}{6} = 0.5$
