### **Data Type**

- **Quantitative Data** (Numerical Data)
    - <font color=Blue>Continuous Data</font>: Values can take any value within a range(e.g. Temperature; Height; Income). Continuous data is ideal for modelling with regression and for computing precise statistical measures.
    - <font color=Blue>Discrete Data</font>: Values can only take specific, countable numbers(e.g. Number of customers; tasks). Discrete data is often used with Poisson or Binomial statistical models.
- **Qualitative Data** (Categorical Data)
    - <font color=Blue>Nominal Data</font>: Categories have no intrinsic ordering (e.g., gender, nationality, colour, product type).
Nominal data is used in frequency tables, bar charts, and statistical tests such as the chi-square test for independence.
    - <font color=Blue>Ordinal Data</font>: Categories possess a meaningful order, but the differences between levels are not numerically consistent (e.g., satisfaction levels such as Poor < Fair < Good < Excellent; education level).
Ordinal data is common in surveys and behavioural research and typically analysed using rank-based or non-parametric methods such as the Mann–Whitney or Wilcoxon tests.
