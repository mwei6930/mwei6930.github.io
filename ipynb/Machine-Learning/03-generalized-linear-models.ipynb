## **Generalized Linear Models**
### **Linear Regression**
Linear regression models the relationship between input variables and a numeric target by fitting a line (or hyperplane) that minimizes the sum of squared prediction errors using the least squares method.
#### **Simple(Bivariate) Linear Regression**



Simple linear regression considers only one input feature and models the relationship between the independent variable $x$ and the dependent variable $y$ using a straight line: $\hat{y} = b_0 + b_1 x$. Where $b_0$ is intercept and $b_1$ is slope  

The goal is to approximate the relationship between $x$ and $y$ and use the fitted line to make predictions for new data points.

#### **Multiple Linear Regression**



Multiple linear regression extends this idea to multiple input features: $\hat{y} = b_0 + b_1 x_1 + b_2 x_2 + \cdots + b_p x_p$.

Instead of a line, the model becomes a **plane (in 2D)** or a **hyperplane (in higher dimensions)**.

#### **Model Fitting**
There are many possible lines that can fit the data. The best one is chosen by minimizing the prediction error.

- Residual (error): $\epsilon_i = y_i - \hat{y}_i$
- Objective function (Sum of Squared Errors, SSE): $SSE = \sum (y_i - \hat{y}_i)^2$

The optimal model is obtained by minimizing SSE using the **least squares method**.

#### **Evaluation Metrics**

- **R² (coefficient of determination)**:$R^2 = \frac{SSR}{SST}$
- Range: $0 \sim 1$ (higher is better)  
- $R^2 = 1$: perfect fit  
- $R^2 \approx 0$: poor fit  

Other commonly used metrics: MAE (Mean Absolute Error), MSE (Mean Squared Error), RMSE (Root Mean Squared Error)
